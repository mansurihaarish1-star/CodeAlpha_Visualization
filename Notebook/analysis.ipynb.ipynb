```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\haarish\OneDrive\Desktop\SalesDataset.csv")
```

```python
df.head()
df.info()
df.describe()
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
sns.set_style("whitegrid")
```

```python

# Handle missing values first
df["Units_Sold"].fillna(df["Units_Sold"].mean(), inplace=True)

# ✅ Create Total_Sales column (this is the fix)
df["Total_Sales"] = df["Units_Sold"] * df["Unit_Price"]

print(df.head())
```

```python
import os

# Create folder if it doesn't exist
os.makedirs("images", exist_ok=True)
product_sales = df.groupby("Product")["Total_Sales"].sum().sort_values()

product_sales.plot(kind="barh")
plt.title("Total Sales by Product")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.savefig("images/sales_by_product.png", bbox_inches='tight')
plt.close()
```

```python
import os

# Create folder if it doesn't exist
os.makedirs("images", exist_ok=True)
region_sales = df.groupby("Region")["Total_Sales"].sum()

region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.savefig("images/sales_by_Region.png", bbox_inches='tight')
plt.close()
```

```python
import os

# Create folder if it doesn't exist
os.makedirs("images", exist_ok=True)
year_sales = df.groupby("Year")["Total_Sales"].sum()

year_sales.plot(kind="line")
plt.title("Sales Trend Over Years")
plt.savefig("images/sales_by_Trend.png", bbox_inches='tight')
plt.close()
```

```python
import os

# Create folder if it doesn't exist
os.makedirs("images", exist_ok=True)
plt.hist(df["Units_Sold"], bins=20)
plt.title("Units Sold Distribution")
plt.savefig("images/sales_by_Distribution.png", bbox_inches='tight')
plt.close()
```

```python
import os

# Create folder if it doesn't exist
os.makedirs("images", exist_ok=True)
plt.scatter(df["Unit_Price"], df["Total_Sales"])
plt.xlabel("Unit Price")
plt.ylabel("Total Sales")
plt.title("Price vs Sales")
plt.savefig("images/price_vs_sales_scatter.png", bbox_inches='tight', dpi=300)
plt.close()
```

```python
import os

# Create folder if it doesn't exist
os.makedirs("images", exist_ok=True)
numeric_df = df.select_dtypes(include=['number'])

sns.heatmap(numeric_df.corr(), annot=True)
plt.savefig("images/price_vs_sales_heatmap.png", bbox_inches='tight', dpi=300)
plt.close()
```
